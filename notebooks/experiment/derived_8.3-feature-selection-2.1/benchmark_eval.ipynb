# Derived 8.3 Feature Selection 2.1 — Reused Benchmark Evaluation

**Purpose.** This notebook is the explicit, post-freeze entry point for the established derived-8.3 Washington 2023–2025 benchmark.

**Method.** It can invoke only the saved benchmark runner, which verifies the complete development freeze before importing benchmark-data code, and otherwise displays already generated benchmark artifacts.

**Result.** Execution is opt-in and disabled by default. A confirmed run evaluates V0 and exactly one predeclared challenger without rewriting development artifacts.

**Interpretation.** This period is a reused retrospective project benchmark—not an untouched holdout—and can never support an unbiased-SOTA or external-generalization claim.


```python
from pathlib import Path
import json
import os
import subprocess
import sys

import pandas as pd
from IPython.display import Markdown, display


def locate_experiment() -> Path:
    for base in (Path.cwd(), *Path.cwd().parents):
        direct = base / "experiment" / "derived_8.3-feature-selection-2.1"
        nested = base / "notebooks" / "experiment" / "derived_8.3-feature-selection-2.1"
        if direct.is_dir():
            return direct.resolve()
        if nested.is_dir():
            return nested.resolve()
    raise FileNotFoundError("Cannot locate derived_8.3-feature-selection-2.1")


EXP_DIR = locate_experiment()
NOTEBOOKS_DIR = EXP_DIR.parents[1]
print(f"Experiment: {EXP_DIR}")
```

## Explicit post-freeze benchmark command

**Purpose.** Evaluate the frozen model IDs exactly once on the reused benchmark after development is complete.

**Method.** Set `MDR_CONFIRM_DERIVED_8_3_BENCHMARK=1`; the cell then calls `run_benchmark.py --confirm-benchmark --device cuda --workers 4`. With no opt-in it only prints the command.

**Result.** The runner rejects a missing or mismatched freeze before any test-data module is imported, validates historical row alignment, and writes only benchmark artifacts and the benchmark report.

**Interpretation.** Benchmark observations cannot feed back into feature identity, beta, routing, or architecture. Any motivated change belongs in a new experiment version.

```python
run_benchmark = os.environ.get(
    "MDR_CONFIRM_DERIVED_8_3_BENCHMARK", "0"
) == "1"
benchmark_command = [
    sys.executable,
    str(EXP_DIR / "run_benchmark.py"),
    "--confirm-benchmark",
    "--device",
    "cuda",
    "--workers",
    "4",
]
if run_benchmark:
    subprocess.run(benchmark_command, cwd=NOTEBOOKS_DIR, check=True)
else:
    print("Opt-in disabled. Explicit retrospective benchmark command:")
    print(" ".join(benchmark_command))
```

## Benchmark metrics and provenance checks

**Purpose.** Present the frozen challenger, 1.3-lite V0, and saved historical Model 16 comparison on identical benchmark rows.

**Method.** The cell reads the overall, station, month, station-year, alignment, paired-bootstrap, and manifest artifacts exactly as written by the benchmark runner.

**Result.** A completed run records 8,396-row historical-label alignment, the R² margin, paired station-year uncertainty, robustness guards, and retrospective provenance flags.

**Interpretation.** A nominal score increase below the 0.003 R² reproducibility margin is not a new project benchmark SOTA.

```python
def show_csv(relative_path: str, rows: int = 50) -> None:
    path = EXP_DIR / relative_path
    if path.exists():
        display(pd.read_csv(path).head(rows))
    else:
        print(f"Pending artifact: {path.relative_to(EXP_DIR)}")


def show_json(relative_path: str) -> None:
    path = EXP_DIR / relative_path
    if path.exists():
        display(json.loads(path.read_text(encoding="utf-8")))
    else:
        print(f"Pending artifact: {path.relative_to(EXP_DIR)}")


benchmark_stage = "artifacts/benchmark"
show_csv(f"{benchmark_stage}/metrics_overall.csv")
show_csv(f"{benchmark_stage}/metrics_by_station.csv")
show_csv(f"{benchmark_stage}/metrics_by_month.csv")
show_csv(f"{benchmark_stage}/metrics_by_station_year.csv")
show_json(f"{benchmark_stage}/historical_alignment.json")
show_json(f"{benchmark_stage}/paired_bootstrap_intervals.json")
show_json(f"{benchmark_stage}/benchmark_manifest.json")
```

## Disclosed project-benchmark claim

**Purpose.** Display the final claim scope and eligibility flags without weakening the experiment's interpretation boundary.

**Method.** The cell renders both the machine-readable `benchmark_claim.json` and generated `BENCHMARK_RESULTS.md` when present.

**Result.** Every verdict fixes `retrospective_test` and `benchmark_reused` to true, while `unbiased_sota_eligible` and `unbiased_generalization_claim_eligible` remain false.

**Interpretation.** A passing result establishes only a project SOTA on the reused derived-8.3 Washington benchmark, pending confirmation on the forthcoming ECE sensor deployment.

```python
show_json(f"{benchmark_stage}/benchmark_claim.json")
benchmark_report = EXP_DIR / "BENCHMARK_RESULTS.md"
display(Markdown(benchmark_report.read_text(encoding="utf-8")))
```
