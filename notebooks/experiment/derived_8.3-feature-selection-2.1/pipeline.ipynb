# Derived 8.3 Feature Selection 2.1 — Development Pipeline

**Purpose.** This notebook is the canonical presentation entry point for the evaluation-first 2.1 development experiment.

**Method.** It invokes only the saved, version-local runner and displays artifacts written by that runner. Feature selection, fitting, decisions, and report generation all remain in tracked Python modules rather than notebook memory.

**Result.** The expensive CUDA run is deliberately opt-in so opening or executing this notebook cannot start it accidentally. Until that run is completed, artifact panels report that results are pending.

**Interpretation.** Only 2017–2022 development evidence may select the configuration. The separate benchmark notebook is the sole route to the reused 2023–2025 benchmark.


```python
from pathlib import Path
import os
import subprocess
import sys

import pandas as pd
from IPython.display import Markdown, display


def locate_experiment() -> Path:
    for base in (Path.cwd(), *Path.cwd().parents):
        direct = base / "experiment" / "derived_8.3-feature-selection-2.1"
        nested = base / "notebooks" / "experiment" / "derived_8.3-feature-selection-2.1"
        if direct.is_dir():
            return direct.resolve()
        if nested.is_dir():
            return nested.resolve()
    raise FileNotFoundError("Cannot locate derived_8.3-feature-selection-2.1")


EXP_DIR = locate_experiment()
NOTEBOOKS_DIR = EXP_DIR.parents[1]
ARTIFACTS = EXP_DIR / "artifacts"
print(f"Experiment: {EXP_DIR}")
```

## Canonical development command

**Purpose.** Run every development stage, including global selection, station diagnostics, causal MoE diagnostics, the immutable development freeze, and generated reports.

**Method.** Set `MDR_RUN_DERIVED_8_3_FS21=1` before execution. The cell then calls `run_all.py --device cuda --workers 4`; without the variable it only prints the exact command.

**Result.** A successful run creates the stage journal, completion markers, OOF ledgers, machine-readable decisions, and `development_freeze.json`.

**Interpretation.** This command is development-only. It never invokes the benchmark and its development import audit rejects access to `test.csv`.

```python
run_development = os.environ.get("MDR_RUN_DERIVED_8_3_FS21", "0") == "1"
development_command = [
    sys.executable,
    str(EXP_DIR / "run_all.py"),
    "--device",
    "cuda",
    "--workers",
    "4",
]
if run_development:
    subprocess.run(development_command, cwd=NOTEBOOKS_DIR, check=True)
else:
    print("Opt-in disabled. Canonical command:")
    print(" ".join(development_command))
```

## Frozen development evidence

**Purpose.** Present the global promotion, beta, and final consensus decisions produced by the saved pipeline.

**Method.** The cell reads the generated CSV and JSON artifacts verbatim; it performs no ranking, model selection, or metric recomputation.

**Result.** After a full run, the tables expose the primary-risk comparison and the machine-readable winner. Before then, each missing artifact is identified explicitly.

**Interpretation.** V0 remains the winner unless the configured paired robustness gates pass; a failed candidate may appear diagnostically but is not promotable.

```python
import json


def show_csv(relative_path: str, rows: int = 30) -> None:
    path = EXP_DIR / relative_path
    if path.exists():
        display(pd.read_csv(path).head(rows))
    else:
        print(f"Pending artifact: {path.relative_to(EXP_DIR)}")


def show_json(relative_path: str) -> None:
    path = EXP_DIR / relative_path
    if path.exists():
        display(json.loads(path.read_text(encoding="utf-8")))
    else:
        print(f"Pending artifact: {path.relative_to(EXP_DIR)}")


show_csv("artifacts/development/stages/07_global_decision/candidate_metrics.csv")
show_json("artifacts/development/global_promotion_decision.json")
show_json("artifacts/development/stages/08_beta_decision/beta_decision.json")
show_json("artifacts/development/candidate_features.json")
show_json("development_freeze.json")
```
