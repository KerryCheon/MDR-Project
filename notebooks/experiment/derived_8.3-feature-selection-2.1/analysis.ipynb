# Derived 8.3 Feature Selection 2.1 — Stability and Error Analysis

**Purpose.** This notebook presents the saved feature-stability, year, month, and station diagnostics for the 2.1 development period.

**Method.** Every table is loaded from artifacts generated by the version-local scripts; no feature selection, metric fitting, or post-hoc filtering occurs here.

**Result.** Executing the notebook before the canonical run is safe and reports pending artifacts. Afterward it provides a reproducible view of all diagnostic evidence.

**Interpretation.** Year-dependent utility is treated as a stability question, and all nine Washington stations remain in scope regardless of performance.


```python
from pathlib import Path
import json

import pandas as pd
from IPython.display import display


def locate_experiment() -> Path:
    for base in (Path.cwd(), *Path.cwd().parents):
        direct = base / "experiment" / "derived_8.3-feature-selection-2.1"
        nested = base / "notebooks" / "experiment" / "derived_8.3-feature-selection-2.1"
        if direct.is_dir():
            return direct.resolve()
        if nested.is_dir():
            return nested.resolve()
    raise FileNotFoundError("Cannot locate derived_8.3-feature-selection-2.1")


def show_csv(exp_dir: Path, relative_path: str, rows: int = 40) -> None:
    path = exp_dir / relative_path
    if path.exists():
        display(pd.read_csv(path).head(rows))
    else:
        print(f"Pending artifact: {path.relative_to(exp_dir)}")


def show_json(exp_dir: Path, relative_path: str) -> None:
    path = exp_dir / relative_path
    if path.exists():
        display(json.loads(path.read_text(encoding="utf-8")))
    else:
        print(f"Pending artifact: {path.relative_to(exp_dir)}")


EXP_DIR = locate_experiment()
print(f"Experiment: {EXP_DIR}")
```

## Feature identity and cross-origin stability

**Purpose.** Inspect which predictors survive the frozen pruning method and whether support changes across 2020, 2021, and 2022.

**Method.** The saved rank summaries, direct/progressive overlap, exact-count consensus support, origin lists, and stability table are displayed without recomputation.

**Result.** The full run records selection frequency, percentile ranks, Jaccard overlap, rank correlation, V0 overlap, and late-year gains or losses.

**Interpretation.** Stable consensus—not any single year's list—defines the final feature identity, so 2021–2022 evidence informs the model without creating year-specific models.

```python
show_csv(
    EXP_DIR,
    "artifacts/development/stages/05_robust_candidate_generation/feature_rank_summary.csv",
)
show_csv(
    EXP_DIR,
    "artifacts/development/stages/04_path_screen/direct_progressive_overlap.csv",
)
show_csv(
    EXP_DIR,
    "artifacts/development/stages/09_consensus/consensus_support.csv",
)
show_csv(
    EXP_DIR,
    "artifacts/development/stages/09_consensus/stability_summary.csv",
)
show_json(
    EXP_DIR,
    "artifacts/development/stages/09_consensus/origin_lists.json",
)
```

## Station, month, and fit-window behavior

**Purpose.** Diagnose whether weak performance reflects compact-list selection failure, current-input limitations, low target variance, or uncertainty.

**Method.** The cell displays all-station classifications, station and month metrics, paired selected-versus-all intervals, transition-month highlights, and the fixed-2017–2019 versus expanding-window diagnostic.

**Result.** Each subgroup includes target dispersion beside prediction metrics, with zero-variance R² recorded as `NaN` and an explicit reason.

**Interpretation.** These diagnostics identify follow-up experiments but cannot prune stations, alter the selected list, or promote a different fit window.

```python
stage_10 = "artifacts/development/stages/10_station_temporal_diagnostics"
show_csv(EXP_DIR, f"{stage_10}/station_sufficiency_classification.csv")
show_csv(EXP_DIR, f"{stage_10}/selected_vs_all_station_intervals.csv")
show_csv(EXP_DIR, f"{stage_10}/metrics_by_station.csv")
show_csv(EXP_DIR, f"{stage_10}/metrics_by_month.csv")
show_csv(EXP_DIR, f"{stage_10}/metrics_by_station_year.csv")
show_csv(EXP_DIR, f"{stage_10}/transition_month_highlights.csv")
show_csv(EXP_DIR, f"{stage_10}/fit_window_metrics.csv")
```

```python
show_csv(EXP_DIR, f"{stage_10}/station_input_diagnostics.csv")
show_csv(EXP_DIR, f"{stage_10}/station_month_residual_distribution.csv")
show_csv(EXP_DIR, f"{stage_10}/station_month_climatology.csv")
show_json(EXP_DIR, f"{stage_10}/correlation_components.json")
show_csv(EXP_DIR, f"{stage_10}/correlation_joint_permutation.csv")
show_csv(EXP_DIR, f"{stage_10}/feature_family_summary.csv")
```
