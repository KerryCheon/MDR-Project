# Derived 8.3 Feature Selection 2.1 — Causal MoE Analysis

**Purpose.** This notebook presents the causal mixture-of-experts diagnostics required even when global feature selection fails.

**Method.** It displays the saved shared-backbone, historical-specialist, router, and regime-delta artifacts generated under the same OOF geometry as the global comparison.

**Result.** Before development execution, panels report pending files; after execution they show the complete ablation and machine-readable MoE verdict.

**Interpretation.** MoE can become the frozen benchmark challenger only after both the global gate and the stricter MoE promotion gate pass.


```python
from pathlib import Path
import json

import pandas as pd
from IPython.display import display


def locate_experiment() -> Path:
    for base in (Path.cwd(), *Path.cwd().parents):
        direct = base / "experiment" / "derived_8.3-feature-selection-2.1"
        nested = base / "notebooks" / "experiment" / "derived_8.3-feature-selection-2.1"
        if direct.is_dir():
            return direct.resolve()
        if nested.is_dir():
            return nested.resolve()
    raise FileNotFoundError("Cannot locate derived_8.3-feature-selection-2.1")


def show_csv(exp_dir: Path, relative_path: str, rows: int = 50) -> None:
    path = exp_dir / relative_path
    if path.exists():
        display(pd.read_csv(path).head(rows))
    else:
        print(f"Pending artifact: {path.relative_to(exp_dir)}")


def show_json(exp_dir: Path, relative_path: str) -> None:
    path = exp_dir / relative_path
    if path.exists():
        display(json.loads(path.read_text(encoding="utf-8")))
    else:
        print(f"Pending artifact: {path.relative_to(exp_dir)}")


EXP_DIR = locate_experiment()
print(f"Experiment: {EXP_DIR}")
```

## Shared-backbone causal matrix and router behavior

**Purpose.** Determine whether any MoE loss comes from missing global information, hard-routing/sample fragmentation, or the historical specialist-list collapse.

**Method.** The cell displays causal-ablation metrics plus target-free router population, centroid-drift, and manifest artifacts from stage 11.

**Result.** Each arm uses the same row-aligned OOF ledger, while router preprocessing is fitted only on outer-training rows and aligned only for reporting.

**Interpretation.** Comparisons among single-global and shared hard-expert arms isolate routing effects before any regime-specific features are considered.

```python
stage_11 = "artifacts/development/stages/11_moe_causal_matrix"
show_csv(EXP_DIR, f"{stage_11}/causal_ablation_metrics.csv")
show_csv(EXP_DIR, f"{stage_11}/router_regime_populations.csv")
show_csv(EXP_DIR, f"{stage_11}/router_drift.csv")
show_json(EXP_DIR, f"{stage_11}/causal_matrix_manifest.json")
```

## Regime-delta coverage and incremental signal

**Purpose.** Test whether unused predictors add stable regime-specific signal after preserving the complete strongest shared backbone.

**Method.** The cell displays saved conditional rankings, regime metrics, and the final frozen expert feature specification for delta counts 0, 5, 10, and 15.

**Result.** Each regime records population, station/origin coverage, target dispersion, rank stability, paired improvement, and the reason for a zero-delta fallback.

**Interpretation.** A nonzero delta represents incremental signal only when its paired interval and worst-station guard pass; insufficient coverage or unstable evidence leaves the shared backbone intact.

```python
stage_12 = "artifacts/development/stages/12_regime_delta_moe_decision"
show_csv(EXP_DIR, f"{stage_12}/regime_delta_rankings.csv")
show_csv(EXP_DIR, f"{stage_12}/regime_metrics.csv")
show_json(EXP_DIR, f"{stage_12}/frozen_moe_features.json")
```
