# Evaluate frozen 2.2 feature sets

This notebook summarizes validation or final metrics created by `run_eval.py`. The held-out test requires the explicit `--confirm-final` command and is never triggered by notebook execution.


## Locate evaluation artifacts

Choose validation or final-test results explicitly. The default reads validation metrics and reports when selection or evaluation has not run yet.

```python
from pathlib import Path
import json

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
for candidate in [PROJECT_ROOT, *PROJECT_ROOT.parents]:
    if (candidate / "data" / "splits").is_dir():
        PROJECT_ROOT = candidate
        break
EXP_DIR = PROJECT_ROOT / "notebooks/experiment/derived_8.2-feature-selection-2.2"
USE_FINAL_TEST = False
EVAL_DIR = EXP_DIR / "artifacts/final" / ("final_test_eval" if USE_FINAL_TEST else "validation_eval")
print("Evaluation directory:", EVAL_DIR)
```

## Compare global and shared-plus-delta performance

Display overall metrics and the immutable gate manifest. Per-year and per-station CSVs in the same directory provide the robustness breakdown.

```python
metrics_path = EVAL_DIR / "metrics_summary.csv"
manifest_path = EVAL_DIR / "evaluation_manifest.json"
if metrics_path.exists():
    metrics = pd.read_csv(metrics_path)
    display(metrics.sort_values(["dataset", "beta", "R2"], ascending=[True, True, False]))
    if manifest_path.exists():
        display(json.loads(manifest_path.read_text()))
else:
    print("Run run_selection.py, then run_eval.py, to generate evaluation artifacts.")
```
