```python
def get_metrics_dict(y_true, y_pred, prefix=""):
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()

    err = y_true - y_pred
    ae = np.abs(err)

    r2 = float(r2_score(y_true, y_pred))
    mae = float(mean_absolute_error(y_true, y_pred))
    rmse = float(root_mean_squared_error(y_true, y_pred))

    bias = float(np.mean(err))
    ubrmse = float(np.std(err))

    q_err = np.quantile(err, [0.05, 0.25, 0.50, 0.75, 0.95])

    return {
        f"{prefix}n": int(len(y_true)),
        f"{prefix}r2": r2,
        f"{prefix}mae": mae,
        f"{prefix}rmse": rmse,
        f"{prefix}ubrmse": ubrmse,
        f"{prefix}bias": bias,
        f"{prefix}med_ae": float(np.median(ae)),
        f"{prefix}p90_ae": float(np.quantile(ae, 0.90)),
        f"{prefix}q05_err": float(q_err[0]),
        f"{prefix}q50_err": float(q_err[2]),
        f"{prefix}q95_err": float(q_err[4]),
    }

def neat_print(metrics):
    print(f"{'METRIC':<15} | {'VALUE':<10}")
    print("-" * 28)
    for k, v in metrics.items():
        val_str = f"{v:,}" if isinstance(v, int) else f"{v:+.5f}"
        print(f"{k:<15} | {val_str:<10}")
```

### Split Strategy
- **Training set:**  
  Two stations, early time period  
- **Validation set:**  
  Same stations as training, held-out **future dates** (temporal holdout)
- **Test set:**  
  One completely unseen station (station-level holdout)

### Motivation
- Validation evaluates **temporal generalization** on known stations
- Test evaluates **spatial generalization** to an unseen station
- This avoids spatial leakage while preserving sufficient training data

```python
print("=== SPLIT SUMMARY ===")

def split_summary(name, d):
    print(f"\n{name.upper()}")
    print(f"  rows:     {len(d)}")
    print(f"  stations: {sorted(d['station_id'].unique().tolist())}")
    if "date" in d.columns:
        print(f"  date range: {d['date'].min()} -- {d['date'].max()}")

split_summary("train", train_df)
split_summary("val", val_df)
split_summary("test", test_df)

print("\n=== LEAKAGE CHECK ===")
print("train ∩ test:", sorted(set(train_df.station_id) & set(test_df.station_id)))
print("val   ∩ test:", sorted(set(val_df.station_id) & set(test_df.station_id)))

print("\n-- split locked --")
```

```python
def variance_by_bins(df, q=8, by="y_true"):
    out = df.copy()
    out["bin"] = pd.qcut(out[by], q=q, duplicates="drop")
    var_bin = out.groupby("bin")["residual"].var()
    mae_bin = out.groupby("bin")["abs_error"].mean()
    cnt_bin = out.groupby("bin")["residual"].size()
    return var_bin, mae_bin, cnt_bin

q = 8

var_no_w, mae_no_w, cnt_no_w = variance_by_bins(df_no_w, q=q, by="y_true")
var_w, mae_w, cnt_w = variance_by_bins(df_w, q=q, by="y_true")

print("Unweighted counts per bin:\n", cnt_no_w)
print("\nWeighted counts per bin:\n", cnt_w)
```

```python
def binned_residual_boxplot(df, q=8, by="y_true", title=""):
    tmp = df.copy()
    tmp["bin"] = pd.qcut(tmp[by], q=q, duplicates="drop")
    groups = [g["residual"].values for _, g in tmp.groupby("bin")]
    labels = [str(i) for i in range(len(groups))]

    plt.figure(figsize=(10, 4))
    plt.boxplot(groups, labels=labels, showfliers=False)
    plt.title(title)
    plt.xlabel("Bin index (low -> high y_true)")
    plt.ylabel("Residual")
    plt.show()
```

```python
def _clean_for_statsmodels(residuals, X):
    r = np.asarray(residuals).ravel()
    Xn = np.asarray(X)

    ok_r = np.isfinite(r)
    ok_X = np.isfinite(Xn).all(axis=1)
    mask_ok = ok_r & ok_X

    r_clean = r[mask_ok]
    X_clean = Xn[mask_ok]

    return r_clean, X_clean, mask_ok
```

```python
residuals_no_w = np.asarray(y_test_d_no_w).ravel() - np.asarray(pred_drift_test_no_w).ravel()
X_sm_raw = sm.add_constant(X_test_d)
resid_no_w_c, X_sm_no_w_c, mask_no_w = _clean_for_statsmodels(residuals_no_w, X_sm_raw)

print(f"No Weights: kept {mask_no_w.sum()} / {len(mask_no_w)} rows (dropped {len(mask_no_w) - mask_no_w.sum()})")

model_no_w = sm.OLS(resid_no_w_c, X_sm_no_w_c).fit()

labels = ['LM Statistic', 'LM-Test p-value', 'F-Statistic', 'F-Test p-value']

bp_test_no_w = sms.het_breuschpagan(model_no_w.resid, model_no_w.model.exog)
print("=== Breusch-Pagan Test (No Weights) ===")
print(dict(zip(labels, bp_test_no_w)))

white_test_no_w = sms.het_white(model_no_w.resid, model_no_w.model.exog)
print("\n=== White Test (No Weights) ===")
print(dict(zip(labels, white_test_no_w)))
```

```python
residuals_w = np.asarray(y_test_d_w).ravel() - np.asarray(pred_drift_test_w).ravel()
X_sm_raw = sm.add_constant(X_test_d)
resid_w_c, X_sm_w_c, mask_w = _clean_for_statsmodels(residuals_w, X_sm_raw)

print(f"Weighted: kept {mask_w.sum()} / {len(mask_w)} rows (dropped {len(mask_w) - mask_w.sum()})")

model_w = sm.OLS(resid_w_c, X_sm_w_c).fit()

bp_test_w = sms.het_breuschpagan(model_w.resid, model_w.model.exog)
print("=== Breusch-Pagan Test (Weighted) ===")
print(dict(zip(labels, bp_test_w)))

white_test_w = sms.het_white(model_w.resid, model_w.model.exog)
print("\n=== White Test (Weighted) ===")
print(dict(zip(labels, white_test_w)))
```

### Conclusion

I dug into the heteroskedasticity properly instead of just eyeballing the residual plot.

First, the formal tests. Both Breusch–Pagan and White tests came back insanely significant for both weighted and unweighted models. So yes, heteroskedasticity is real. It’s not subtle. And importantly, weighting barely changed the result. That tells me this isn’t just simple variance scaling.

Then I looked at variance across soil moisture quantile bins. The biggest spike in residual variance isn’t at the highest moisture. It’s in the lower-mid range, roughly around 0.136 to 0.2. That’s interesting. It’s not monotonic. It’s not a clean funnel shape. It’s a specific region where the model struggles most.

The scale–location plots confirm that too. I don’t see a smooth widening of residuals as predictions increase. Instead, I see clusters. One cluster at low moisture. Another at high moisture. And a messy transition zone in between.

When I tried to model the variance itself using a temporal holdout split, the R² was negative. That means the variance pattern does not generalize forward in time. So this isn’t just “predictable noise” that I can learn and weight away. It’s unstable.

Putting that together:
- The heteroskedasticity is real.
- It is not purely magnitude-driven.
- Weighting helps slightly but does not solve the structure.
- The highest variance is concentrated in a specific moisture regime.
- The variance pattern shifts over time.

So this does not look like classic heteroskedasticity that I fix with better loss functions.

It looks like I am trying to model multiple physical regimes with a single function.

--- 

**My Interpretation**

Low moisture behaves differently from higher moisture. And that transition region in the lower-mid range is where the physics change. That is exactly where the model struggles most.

So this isn’t just a noise problem.

**It’s a regime problem.**

- 3 (0.2–0.234)

**Best (least bad)**:

_Ergo:_

- Dry: <= 0.20
- Transition / messy zone: (0.20, 0.313] `->` interval
- Wet: > 0.313
