## 1. Environment and evidence
The existing uv scientific environment and pinned export dependencies support a fully reproducible build. The evidence extractor checks matched seeds and station coverage, prints every displayed value with provenance, and verifies that the numerical results and station memberships match poster 1.0.

```python
from pathlib import Path
import sys
import subprocess
from importlib import metadata

cwd = Path.cwd().resolve()
repo = next(parent for parent in (cwd, *cwd.parents) if (parent / "notebooks/uv.lock").is_file())
poster_root = repo / "writeup2/poster-summer-1.1"
requirements = poster_root / "requirements.txt"
missing = []
for line in requirements.read_text().splitlines():
    if not line or line.startswith("#"):
        continue
    package, version = line.split("==")
    try:
        installed = metadata.version(package)
    except metadata.PackageNotFoundError:
        installed = None
    if installed != version:
        missing.append(line)
if missing:
    subprocess.run(["uv", "pip", "install", "--python", sys.executable, *missing], check=True, timeout=180)
sys.path.insert(0, str(poster_root / "scripts"))
import poster_data

data = poster_data.load_evidence()
result_table = poster_data.write_evidence(data)
print("Evidence validation passed: 30 paired temporal seeds; 5 seeds × 7 held-out stations.")
print("Regime names changed; fitted memberships and underlying results match poster 1.0.")
```
